```python
import pandas as pd
import datetime as dt

input_data = pd.DataFrame(
    {
        'Roll No':[1,1,2,2,1,2,2],
        'Subject':['DS','Java','DS','Java','DS','Java','Java'],
        'Marks':[25,40,30,26,40,50,58],
        'Date':[
            "01-01-23",
            "02-01-23",
            "01-01-23",
            "02-01-23",
            "15-01-23",
            "15-01-23",
            "20-01-23"]
    }
)

```

```python
input_data["Date"] = pd.to_datetime(input_data["Date"], format="%d-%m-%y")
input_data = input_data.sort_values("Date",ascending=False,kind="stable")
keys = ["Roll No","Subject"]




```

```python
input_data['Attempt'] = input_data.groupby(keys).cumcount() + 1
```

```python
input_data
```

```python
latest_marks = (
    input_data[input_data["Attempt"]<=3]
    .pivot(index=keys,columns="Attempt",values="Marks")
    .reindex(columns=[1,2,3])
    .fillna(0)
    .astype(int)
    .rename(columns={1:'M1',2:'M2',3:'M3'})
    
)
```

```python
latest_marks
```

```python
latest_dates = input_data.groupby(keys,as_index=False)["Date"].first()
```

```python
output = latest_marks.reindex().merge(latest_dates,on=keys)
output['Date'] = output['Date'].dt.strftime("%d-%m-%y")
output =output[['Roll No','Subject','M1','M2','M3','Date']] 
```
